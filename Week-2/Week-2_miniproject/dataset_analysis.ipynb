```python
"""
Part B: Download a public CSV (Titanic, Iris, etc.). 
Clean it: handle missing values, drop irrelevant columns. 
pandas: groupby summary, top-10 ranking, filtered subset with 2+ conditions. 
numpy: mean, std deviation, 25th/75th percentile, correlation between two columns. 
"""
```

```python
import pandas as pd
df=pd.read_csv("titanic.csv")
print(df.shape)
print(df.head(n=10))
print(df.describe)
print(df.info())

```

```python
#Clean it: handle missing values, drop irrelevant columns.

df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Age"]=df["Age"].fillna(df["Age"].mean)
#df=df.drop(columns=["cabin"])
print(df.isnull().sum())
df.drop_duplicates()

```

```python
#pandas: groupby summary, top-10 ranking, filtered subset with 2+ conditions. 
print(df["Ticket"].dtype)
df["Age"]=pd.to_numeric(df["Age"],errors="coerce")
df["Ticket"]=pd.to_numeric(df["Ticket"],errors="coerce")
print("Average age of male and female")
summ=df.groupby("Sex")["Age"].mean()
summary=df.groupby("Pclass")["Age"].mean()
#s=df.groupby("Pclass").agg(a=("Sex").mode()[0],b=("Survived").count())
#print(s)
print(summary)
print(summ)
sorted=df.sort_values(by="Fare",ascending=False)
print(sorted.head(n=10))
df["Ticket"]=df["Ticket"].fillna((df["Ticket"].mean()).round(2))
print(df.isnull().sum())
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
#numpy: mean, std deviation, 25th/75th percentile, correlation between two columns. 
import numpy as np
import sys
print(df.columns)
np.set_printoptions(threshold=sys.maxsize)
Age=np.array(pd.to_numeric(df["Age"],errors="coerce"))
data=np.array(df)
print(data[:20,:])
print(np.mean(data[:,5]))
print(np.percentile(data[:,8],75))
print(np.percentile(data[:,8],50))
print(np.correlate(data[:,5],data[:,8]))
```
