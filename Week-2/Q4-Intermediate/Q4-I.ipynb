```python
import pandas as pd
data={"student_id":["101","101","102","102","103","103"],
      "name":["a","a","b","b","c","c"],
      "class":[7,7,9,9,10,10]}
df1=pd.DataFrame(data)
print(df1)
```

```python
datas={"student_id":["101","101","102","102","103","103"],
      "subject":["Maths","science","Maths","science","Maths","science"],
      "score":[70,49,57,98,67,87]}

df2=pd.DataFrame(datas)
print(df2)
```

```python
df1.to_csv("students.csv",index=False)
df2.to_csv("scores.csv",index=False)
```

```python

student_data=pd.read_csv("students.csv")
score_data=pd.read_csv("scores.csv")

print(student_data)
print(score_data)
```

```python
merged_data=pd.merge(student_data,score_data,on="student_id")
print(merged_data)
merged_data=merged_data.drop_duplicates()
```

```python
#Use groupby to find average score per subject and per student.
merged_data.groupby("subject")["score"].mean()
```

```python
res=merged_data.groupby("student_id")["score"].mean()
print(res)
print(res.to_dict())
```

```python
#Add a grade column: 'A' >= 85, 'B' >= 70, 'C' >= 55, 'F' otherwise.
merged_data["grade"]=merged_data["score"].apply(lambda x:"A" if x>=85 else "B" if x>=70 else "C" if x>=55 else "F")
print(merged_data)
```

```python
#Use numpy: mean, standard deviation, 75th percentile of all scores. 
import numpy as np
print(f"Mean:{np.mean(merged_data["score"].values)}")
print(f"Standard_Deviation:{np.std(merged_data["score"].values)}")
print(f"75TH Percentile:{np.percentile(merged_data["score"].values,75)}")
```
