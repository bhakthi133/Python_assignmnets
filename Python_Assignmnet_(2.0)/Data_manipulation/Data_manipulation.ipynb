```python
import pandas as pd
con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Bhak@13",
    database="assignment_db"
)


df = pd.read_sql("SELECT * FROM json_to_sql_table", con)


print(df)

con.close()
```

```python
df["email"] = df["email"].str.split("@")
print(df["email"])
df["email"] = df["email"].str[0]
print(df["email"])
df["email"] = df["email"] + "@gmail.com"
print(df["email"])
```

```python
df["postalZip"] = df["postalZip"].astype(str)
def remove_letters(value):
    result = ""

    for ch in value:
        if ch.isdigit():
            result += ch

    return result

df["postalZip"] = df["postalZip"].apply(remove_letters)

print(df["postalZip"])
df["postalZip"] = df["postalZip"].astype(int)
```

```python

def code_phone(phone):
    number=remove_letters(phone)
    result = ""
    i = 0

    while i < len(number) - 1:
        pair = number[i] + number[i + 1]
        value = int(pair)

        if value < 65:
            result = result + "O"
        else:
            result = result + chr(value)

        i = i + 2

    return result

df["phone"] = df["phone"].apply(code_phone)
df.rename(columns={"phone": "coded_phone_number"}, inplace=True)
```
