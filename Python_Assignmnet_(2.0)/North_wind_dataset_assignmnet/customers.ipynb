```python
from flask import Flask
from flask_sqlalchemy import SQLAlchemy
from sqlalchemy import URL, Text, LargeBinary

app = Flask(__name__)

url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="Bhak@13",  
    host="localhost",
    port=3306,
    database="northwind"
)

app.config["SQLALCHEMY_DATABASE_URI"] = url

db = SQLAlchemy(app)
```

```python

class Customer(db.Model):

    __tablename__ = "customers"

    id = db.Column(db.Integer, primary_key=True, autoincrement=True)
    company = db.Column(db.String(50))
    last_name = db.Column(db.String(50))
    first_name = db.Column(db.String(50))
    email_address = db.Column(db.String(50))
    job_title = db.Column(db.String(50))
    business_phone = db.Column(db.String(25))
    home_phone = db.Column(db.String(25))
    mobile_phone = db.Column(db.String(25))
    fax_number = db.Column(db.String(25))
    address = db.Column(Text)
    city = db.Column(db.String(50))
    state_province = db.Column(db.String(50))
    zip_postal_code = db.Column(db.String(15))
    country_region = db.Column(db.String(50))
    web_page = db.Column(Text)
    notes = db.Column(Text)
    attachments = db.Column(LargeBinary)

with app.app_context():
    with db.engine.connect() as connection:
        print("Database connected success") 
```

```python

@app.route("/customers",methods=["GET"])
def get_customers():
    cus=Customer.query.all()
    return cus

```

```python
from flask import request
@app.route("/customers",methods=["POST"])
def add_customers():
    data=request.get_json()

    customer = Customer(
        company=data.get("company"),
        last_name=data.get("last_name"),
        first_name=data.get("first_name"),
        email_address=data.get("email_address"),
        job_title=data.get("job_title"),
        business_phone=data.get("business_phone"),
        home_phone=data.get("home_phone"),
        mobile_phone=data.get("mobile_phone"),
        fax_number=data.get("fax_number"),
        address=data.get("address"),
        city=data.get("city"),
        state_province=data.get("state_province"),
        zip_postal_code=data.get("zip_postal_code"),
        country_region=data.get("country_region"),
        web_page=data.get("web_page"),
        notes=data.get("notes"),
        attachments=data.get("attachments")
    )

    db.session.add(customer)
    db.session.commit()

    return {"message": "Customer added","id": customer.id }, 201
```
