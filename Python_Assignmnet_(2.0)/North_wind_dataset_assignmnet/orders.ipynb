```python

from flask import Flask, request
from flask_sqlalchemy import SQLAlchemy
from sqlalchemy import URL, Text
from datetime import datetime

app = Flask(__name__)

url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="YOUR_MYSQL_PASSWORD",
    host="localhost",
    port=3306,
    database="northwind"
)

app.config["SQLALCHEMY_DATABASE_URI"] = url

db = SQLAlchemy(app)


# Order Model

class Order(db.Model):

    __tablename__ = "orders"

    id = db.Column(db.Integer, primary_key=True, autoincrement=True)
    employee_id = db.Column(db.Integer)
    customer_id = db.Column(db.Integer)
    order_date = db.Column(db.DateTime)
    shipped_date = db.Column(db.DateTime)
    shipper_id = db.Column(db.Integer)
    ship_name = db.Column(db.String(50))
    ship_address = db.Column(Text)
    ship_city = db.Column(db.String(50))
    ship_state_province = db.Column(db.String(50))
    ship_zip_postal_code = db.Column(db.String(50))
    ship_country_region = db.Column(db.String(50))
    shipping_fee = db.Column(db.Numeric(19, 4))
    taxes = db.Column(db.Numeric(19, 4))
    payment_type = db.Column(db.String(50))
    paid_date = db.Column(db.DateTime)
    notes = db.Column(Text)
    tax_rate = db.Column(db.Float)
    tax_status_id = db.Column(db.Integer)
    status_id = db.Column(db.Integer)




@app.route("/orders", methods=["GET"])
def get_orders():

    orders = Order.query.all()


    return orders, 200




@app.route("/orders", methods=["POST"])
def add_orders():

    data = request.get_json()

    order = Order(
        employee_id=data.get("employee_id"),
        customer_id=data.get("customer_id"),
        order_date=datetime.fromisoformat(data["order_date"]) if data.get("order_date") else None,
        shipped_date=datetime.fromisoformat(data["shipped_date"]) if data.get("shipped_date") else None,
        shipper_id=data.get("shipper_id"),
        ship_name=data.get("ship_name"),
        ship_address=data.get("ship_address"),
        ship_city=data.get("ship_city"),
        ship_state_province=data.get("ship_state_province"),
        ship_zip_postal_code=data.get("ship_zip_postal_code"),
        ship_country_region=data.get("ship_country_region"),
        shipping_fee=data.get("shipping_fee"),
        taxes=data.get("taxes"),
        payment_type=data.get("payment_type"),
        paid_date=datetime.fromisoformat(data["paid_date"]) if data.get("paid_date") else None,
        notes=data.get("notes"),
        tax_rate=data.get("tax_rate"),
        tax_status_id=data.get("tax_status_id"),
        status_id=data.get("status_id")
    )

    db.session.add(order)
    db.session.commit()

    return {"message": "Order added", "id": order.id}, 201


# 3. PUT - Update Order

@app.route("/orders/<int:id>", methods=["PUT"])
def update_orders(id):

    o = db.session.get(Order, id)

    if o is None:
        return {"error": "Order not found"}, 404

    data = request.get_json()

    o.employee_id = data.get("employee_id", o.employee_id)
    o.customer_id = data.get("customer_id", o.customer_id)
    o.order_date = datetime.fromisoformat(data["order_date"]) if data.get("order_date") else o.order_date
    o.shipped_date = datetime.fromisoformat(data["shipped_date"]) if data.get("shipped_date") else o.shipped_date
    o.shipper_id = data.get("shipper_id", o.shipper_id)
    o.ship_name = data.get("ship_name", o.ship_name)
    o.ship_address = data.get("ship_address", o.ship_address)
    o.ship_city = data.get("ship_city", o.ship_city)
    o.ship_state_province = data.get("ship_state_province", o.ship_state_province)
    o.ship_zip_postal_code = data.get("ship_zip_postal_code", o.ship_zip_postal_code)
    o.ship_country_region = data.get("ship_country_region", o.ship_country_region)
    o.shipping_fee = data.get("shipping_fee", o.shipping_fee)
    o.taxes = data.get("taxes", o.taxes)
    o.payment_type = data.get("payment_type", o.payment_type)
    o.paid_date = datetime.fromisoformat(data["paid_date"]) if data.get("paid_date") else o.paid_date
    o.notes = data.get("notes", o.notes)
    o.tax_rate = data.get("tax_rate", o.tax_rate)
    o.tax_status_id = data.get("tax_status_id", o.tax_status_id)
    o.status_id = data.get("status_id", o.status_id)

    db.session.commit()

    return {"message": "Order updated", "id": o.id}, 200



@app.route("/orders/<int:id>", methods=["GET"])
def get_order(id):

    o = db.session.get(Order, id)

    if o is None:
        return {"message": "Invalid order id"}, 404

    return o, 200





if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```
