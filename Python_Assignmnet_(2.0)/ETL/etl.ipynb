```python
"""
ETL Job:
Download the files needed to complete this assignment from here
Go through StackOverflow’s developer survey 2019 dataset.
Submit the solution for the following questions.
1. Find the average age of developers when they wrote their first line of code.
2. Deduce the percentage of developers who knew python in each country.
3. Generate the report for the average salary of developers based on the
continent.
4. Based on this survey, what will be the most desired programming language for the year
2020?
5. Generate the report for the people who code as a hobby based on their gender
and continent.
Note: Please make sure that gender has three categories only MAN, WOMAN,
and OTHERS. If a record has a singular/atomic value as MAN or WOMAN then
only consider it for gender category  MAN and WOMAN respectively, if it has
ambiguous value then consider it for OTHERS.
6. Generate the report for job and career satisfaction of developers based on their
gender and continent.
Note: Please follow the same instructions for the gender category as described in
question-5."""
```

```python
import pandas as pd
df=pd.read_csv("survey_results_public.csv")
#print(df.head)
#print(df.head(n=1))
l=list(df.columns)
l.sort()
print(l)
print(df.columns)
#print(df["Age1stCode"].dtype)
```

```python

df["Age1stCode"] = pd.to_numeric(df["Age1stCode"], errors="coerce")
print(df["Age1stCode"].mean())
```

```python
#2. Deduce the percentage of developers who knew python in each country.
python = df["LanguageWorkedWith"].str.contains("Python", na=False)
print(python.groupby(df["Country"]).mean() * 100)
```

```python
import country_converter as coco

cc = coco.CountryConverter()

df["Continent"] = cc.convert(
    names=df["Country"].fillna("Unknown"),
    to="continent",
    not_found="Unknown"
)
```

```python

print(df["Continent"])
```

```python
df["ConvertedComp"] = pd.to_numeric(df["ConvertedComp"], errors="coerce")

salary = df.groupby("Continent")["ConvertedComp"].mean()

print(salary)
```

```python
#(4) the most desired programming language for 2020, based on the survey;
print(df["LanguageDesireNextYear"])
```

```python
df["list0flanguage"]= df["LanguageDesireNextYear"].str.split(r"[,;/]")
#print(df["list0flanguage"])
list_lan= df["list0flanguage"].explode()
print(list_lan)
print(list_lan.value_counts())
print(list_lan.value_counts().head(n=1))
```

```python
#(5) a report of people who code as a hobby, broken down by gender 
# (categorized strictly as MAN, WOMAN, or OTHERS -- ambiguous/multi-value gender data goes to OTHERS) and continent;
def covert(gen):
    if(gen=="Man"):
        gen="MAN"
    elif(gen=="Woman"):
        gen="WOMEN"
    else:
        gen="OTHERS"
    return gen
#print(df["Gender"])
df["GenderCategory"] = df["Gender"].apply(covert)
print(df["GenderCategory"])
```

```python
hobby_code=df[df["Hobbyist"]=="Yes"]
print(hobby_code.groupby(["GenderCategory","Continent"]).size())
```

```python
#print(df["JobSat"].unique())
job_data = df.groupby(["GenderCategory", "Continent"])
print(job_data["JobSat"].value_counts())
```
